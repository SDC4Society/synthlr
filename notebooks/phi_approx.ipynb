```python
import numpy as np
from matplotlib import pyplot as plt

from synthlr.phi import phi_QMS
```

```python
lams = np.arange(1, 100) * 100
rs = np.arange(1, 11) * 0.1
out = np.zeros(len(lams))

for r in reversed(rs):
    phis = [phi_QMS(int(lam * r), lam) for lam in lams]
    plt.plot(lams, phis, label=f"r={str(r)[:3]}")
    
plt.xlabel(r"$n+\gamma$", fontsize=20)
plt.ylabel(r"$\phi(m, n+\gamma)$", fontsize=20)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.legend(fontsize=15)

# plt.savefig("figs/phi_approx/phi_saturation_various_m.eps", bbox_inches="tight")
plt.show()
```

```python
lam = 1000000
cs = np.arange(1, 101) * 0.01

out = np.zeros(len(cs))

for i in range(len(cs)):
    m = int(lam * cs[i])
    out[i] = phi_QMS(m, lam)

plt.plot(cs, out)
plt.xlabel("c")
plt.ylabel(r"$\phi$")
plt.show()
```

```python
plt.plot(cs, out, label="real")
plt.plot(cs, (1 + cs)**2, label=r"$(1+c)^2$")
plt.plot(cs, 1 + 2*cs, label=r"$1+2c$")
plt.xlabel("c")
plt.ylabel(r"$\phi$")
plt.legend()
plt.show()
```

```python
lams = np.arange(1, 100) * 1000
r = 0.5

plt.axhline(y=(r+1)**2, color='red', linestyle='--', label=r"$(1+r)^2$")

phis = [phi_QMS(int(lam * r), lam) for lam in lams]
plt.plot(lams, phis, label=r"$\phi(m, n+\gamma)$")

plt.xlabel(r"$n+\gamma$", fontsize=20)
# plt.ylabel(r"$\phi$")
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
plt.legend(fontsize=15)

plt.savefig("figs/phi_approx/phi_saturation_r05.eps", bbox_inches="tight")
plt.show()
```
